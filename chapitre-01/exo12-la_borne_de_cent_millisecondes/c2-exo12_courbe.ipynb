# Exercice 12 : cinq tests

```python
import pandas as pd
import matplotlib.pyplot as plt

# Lecture des cinq resultats deja enregistres.
data = pd.read_csv('exo12_cinq_tests.csv')

plt.figure(figsize=(8, 5))
plt.plot(
    data['duree_ms'],
    data['erreur_degres'],
    'o-',
    label='Erreur angulaire',
)
plt.axvline(100, color='gray', linestyle='--', label='Borne 100 ms')
plt.xlabel('Duree (ms)')
plt.ylabel('Erreur (degres)')
plt.title('Erreur d extrapolation')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig('exo12_cinq_tests.png', dpi=150)
plt.show()
```
